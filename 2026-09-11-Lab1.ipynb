# **Problem 1: BFS and DFS Traversal of a Graph**

Problem Statement:
A small graph, shaped like a tree with six nodes (A, B, C, D, E, F) and node A as the root, is given below. Write a Python program that:
1. Stores the graph using an adjacency list (a Python dictionary).
2. Performs a BFS traversal starting from node A.
3. Performs a DFS traversal starting from node A.
4. Prints both traversal orders, and draws two separate tree diagrams — one showing the order in which BFS visits the nodes, and another showing the order in which DFS visits the nodes.

```python
from collections import deque

# Nodes of the graph
nodes = ['A', 'B', 'C', 'D', 'E', 'F']

# Create an empty adjacency list
graph = {}

for node in nodes:
    graph[node] = []

# Take graph connections from the user
print("Enter the connections between the nodes.")
print("Example: A B")
print("Enter 5 connections for the tree:")

for i in range(5):
    u, v = input("Enter connection: ").split()

    graph[u].append(v)
    graph[v].append(u)

# ---------------- BFS ----------------
def bfs(start):
    visited = []
    queue = deque([start])

    while queue:
        node = queue.popleft()

        if node not in visited:
            visited.append(node)

            for neighbour in graph[node]:
                if neighbour not in visited:
                    queue.append(neighbour)

    return visited


# ---------------- DFS ----------------
def dfs(start, visited=None):
    if visited is None:
        visited = []

    visited.append(start)

    for neighbour in graph[start]:
        if neighbour not in visited:
            dfs(neighbour, visited)

    return visited


# Perform traversals
bfs_order = bfs('A')
dfs_order = dfs('A')

# Print adjacency list
print("\nAdjacency List:")
for node in graph:
    print(node, ":", graph[node])

# Print traversal orders
print("\nBFS Traversal:")
print(" -> ".join(bfs_order))

print("\nDFS Traversal:")
print(" -> ".join(dfs_order))


# ---------------- Tree Diagrams ----------------

print("\nBFS Tree Diagram:")
for i, node in enumerate(bfs_order):
    print("   " * i + node)

print("\nDFS Tree Diagram:")
for i, node in enumerate(dfs_order):
    print("   " * i + node)
```

# **Problem 2: The Water Jug Problem (Solved using BFS)**

Problem Statement:
You are given two jugs: Jug A with a capacity of 4 litres, and Jug B with a capacity of 3 litres. Neither jug has any markings on it. Using only the following operations —
(i) fill a jug completely,
(ii) empty a jug completely,
(iii) pour water from one jug into the other until either the first jug is empty or the second jug is full — determine the minimum sequence of steps needed to measure out exactly 2 litres of water in any one jug.


```python
from collections import deque

# Take capacities from the user
capacity_a = int(input("Enter capacity of Jug A: "))
capacity_b = int(input("Enter capacity of Jug B: "))

target = int(input("Enter the amount to measure: "))

# Starting state
start = (0, 0)

# Queue stores (current state, path)
queue = deque()
queue.append((start, []))

# Keep track of visited states
visited = set()
visited.add(start)

solution = None

while queue:
    (a, b), path = queue.popleft()

    # Check if target is reached
    if a == target or b == target:
        solution = path
        solution.append((a, b))
        break

    # All possible operations
    states = [
        ((capacity_a, b), "Fill Jug A"),
        ((a, capacity_b), "Fill Jug B"),
        ((0, b), "Empty Jug A"),
        ((a, 0), "Empty Jug B")
    ]

    # Pour A into B
    amount = min(a, capacity_b - b)
    states.append(((a - amount, b + amount), "Pour A -> B"))

    # Pour B into A
    amount = min(b, capacity_a - a)
    states.append(((a + amount, b - amount), "Pour B -> A"))

    # Add new states to the queue
    for new_state, operation in states:
        if new_state not in visited:
            visited.add(new_state)
            queue.append((new_state, path + [new_state]))

# Print result
if solution:
    print("\nMinimum sequence of steps:")

    print("Step 0: (0, 0)")

    for i, state in enumerate(solution, 1):
        print("Step", i, ":", state)

    print("\nJug A =", solution[-1][0], "litres")
    print("Jug B =", solution[-1][1], "litres")

else:
    print("\nNo solution is possible.")
```

# **Problem 3: Checking Routes Between Cities Using DFS**

Problem Statement:
A small road network connects several cities. Write a Python program using DFS that:
1. Checks whether a route exists between two given cities.
2. Finds all separate groups of cities that are connected to each other but not to any
other group (these are called connected components).

```python
# Create the graph
graph = {}

# Take number of cities
n = int(input("Enter number of cities: "))

# Take city names
print("Enter city names:")
for i in range(n):
    city = input("City: ")
    graph[city] = []

# Take number of roads
roads = int(input("\nEnter number of roads: "))

# Take road connections
print("Enter the roads:")
for i in range(roads):
    city1, city2 = input("Enter two connected cities: ").split()

    graph[city1].append(city2)
    graph[city2].append(city1)


# ---------------- DFS for route ----------------

def route_exists(start, end, visited):
    if start == end:
        return True

    visited.add(start)

    for city in graph[start]:
        if city not in visited:
            if route_exists(city, end, visited):
                return True

    return False


# ---------------- Find Connected Components ----------------

def find_component(city, visited, component):
    visited.add(city)
    component.append(city)

    for neighbour in graph[city]:
        if neighbour not in visited:
            find_component(neighbour, visited, component)


# Check route between two cities
print("\n--- Check Route ---")

start = input("Enter starting city: ")
end = input("Enter destination city: ")

visited = set()

if route_exists(start, end, visited):
    print("Route exists between", start, "and", end)
else:
    print("No route exists between", start, "and", end)


# Find connected components
print("\n--- Connected Components ---")

visited = set()
components = []

for city in graph:
    if city not in visited:
        component = []
        find_component(city, visited, component)
        components.append(component)

for i, component in enumerate(components, 1):
    print("Group", i, ":", component)
```
